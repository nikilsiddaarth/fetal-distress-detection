```python
import pandas as pd
import os

# Load original CTG data
df = pd.read_excel("../data/CTG.xls", sheet_name="Data", header=1)

print("Original shape:", df.shape)
print("\nOriginal columns:")
print(df.columns.tolist())

# Remove completely empty columns
df = df.dropna(axis=1, how="all")

# Remove Excel formatting columns
df = df.loc[:, ~df.columns.astype(str).str.startswith("Unnamed")]

print("\nAfter removing empty/formatting columns:")
print(df.shape)
print(df.columns.tolist())

# Classification/annotation columns that should NOT be used as ML inputs
leakage_columns = [
    "A", "B", "C", "D", "E",
    "AD", "DE", "LD", "FS", "SUSP", "CLASS"
]

# Remove leakage/annotation columns
existing_leakage = [c for c in leakage_columns if c in df.columns]
df = df.drop(columns=existing_leakage)

print("\nRemoved classification columns:")
print(existing_leakage)

# Check target
print("\nTarget distribution:")
print(df["NSP"].value_counts().sort_index())

print("\nFinal columns:")
for i, col in enumerate(df.columns, start=1):
    print(i, ":", col)

print("\nFinal shape:", df.shape)
```
